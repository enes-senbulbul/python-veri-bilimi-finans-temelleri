### 1 - Sayısal Tipler ve Aritmetik (1/5)

Bağlam: Bir elektrik mühendisliği ölçüm aracında, hem doğru akım (DC) hem de alternatif akım (AC) devreleri için temel elektriksel hesaplamalar yapacak bir script yazıyoruz.
Gereksinimler:
1. akim (int) ve gerilim (float) parametrelerini alan bir direnç hesabı (Gerilim / Akım) yap. Akımın sıfır gelmesi durumunu (sıfıra bölünme hatası) basit bir if koşuluyla engelleyip, varsayılan olarak 0.0 döndür.
2. AC devreleri için empedans_1 ve empedans_2 karmaşık sayılarını (complex tipinde, örn: 3 + 4j ) toplayarak eşdeğer empedansı hesapla.
3. Hesaplanan eşdeğer empedansın .real (gerçek) ve .imag (sanal) kısımlarını ayrı ayrı formatlayarak ekrana yazdır. 
4. Python'daki bölme operatörlerinin farkını göstermek için: 100 birimlik gücü 3 farklı alt sisteme eşit dağıt; tam sayı bölmesi (
// ) ile her sisteme düşen payı ve mod alma ( % ) ile arta kalan gücü hesapla.

```python
# 1
def direnc_hesabi(akim, gerilim):
    if abs(akim)<0.05:
        direnc = 0.0
    else:
        direnc = gerilim/akim
    return round(direnc,3)

print(direnc_hesabi(3, 50))
print(direnc_hesabi(0.01, 400))

# 2, 3
empedans_1 = 3 + 4j
empedans_2 = -5j
toplam_empedans = empedans_1 + empedans_2
print("Toplam Empedans:", toplam_empedans)
print("Gerçek kısım:", toplam_empedans.real)
print("Sanal kısım:", toplam_empedans.imag)

# 4
toplam_saniye = 240_874      # 1 saat = 60 dakika = 3600 saniye
dakika_saniye = toplam_saniye % 3600
saat = toplam_saniye // 3600
saniye = dakika_saniye % 60
dakika = dakika_saniye // 60

sep = "-"*10
print("%10s %10s %10s %10s" % ("Toplam(sn)","= Saat","Dakika","Saniye"))
print("%10s %10s %10s %10s" % (sep, sep, sep, sep))
print("%10d %10d %10d %10d" % (toplam_saniye, saat, dakika, saniye))
```

### 2 - Metin Tipi ve Temel String Metotları (4/5)

Bağlam: Bir sunucu altyapısında üretilen, boşlukları fazla bırakılmış ve isimlendirme standartlarına uyulmamış log (günlük) mesajlarını temizleyip anlamlı bir yapıya dönüştüren bir metin ayrıştırıcı (parser) yazıyoruz.
Gereksinimler:
1. Baştaki ve sondaki gereksiz boşlukları strip() metoduyla temizle.
2. Log mesajını seviye ve içerik olarak ayırabilmek için split() kullan.
3. Çıkarılan log seviyesindeki (örneğin [ warning ] ) köşeli parantezleri ve boşlukları temizleyip, standardizasyon için tamamen büyük harfe ( upper() ) çevir.
4. Log içeriğindeki alt çizgileri ( _ ) replace() ile boşluğa çevir ve cümlenin sadece ilk harfini capitalize() ile büyüt.
5. Mesajın sonunda üç nokta ( ... ) olup olmadığını endswith() ile kontrol ederek bir uyarı bayrağı (flag) oluştur. Edge-case olarak boş string gelirse veya hatalı format varsa boş bir sözlük döndür.

```python
test_loglari = [
    "   [ warning ]   db_connection_timeout...   ",
    "[INFO] kullanici_girisi_basarili",   "   \n",  
    "HATALI LOG FORMATI"]

def log_ayristir(ham_log):

    # Her zaman önce edge-case kontrolü yapıyoruz.
    if not ham_log or not ham_log.strip():  # İlk kontrol "None"ı da elemek için, yoksa sadece strip'li olanı yazabilirdik
        print("Uyarı: Boş log girdiniz.")
        return {}

    temiz_log = ham_log.strip()

    if "]" not in temiz_log:
        print("Uyarı: Hatalı log formatı - ']' bulunamadı.")
        return {}

    parts = temiz_log.split("]", 1)  # Sadece 1 defa, ']' karakterinden bölüyor.

    seviye = parts[0].replace("[","").strip().upper()
    ham_mesaj = parts[1].strip()
    okunabilir_mesaj = ham_mesaj.replace("_"," ").capitalize()
    kesilmis_mi = okunabilir_mesaj.endswith("...")

    sonuc = {
        "seviye": seviye,
        "mesaj": okunabilir_mesaj,
        "kesik_mesaj": kesilmis_mi
    }

    return sonuc

for indeks, log_metni in enumerate(test_loglari, 1):    # enumarate() for loopları için çok kullanışlı
    print(f"--- Test #{indeks} ---")
    print(f"Giren: {repr(log_metni)}")  # repr() sayesinde string, kaçış karakterleri de dahil olduğu gibi yazılıyor
    sonuc = log_ayristir(log_metni)
    print(f"Çıkan: {sonuc}\n")
```

### 3 - Boolean, None Nesnesi ve Doğruluk Değeri Kuralları (1/5)

Bağlam: Bir bulut sunucusu yönetim betiği yazıyoruz. Kullanıcıdan gelen çeşitli yapılandırma ayarlarını (IP adresi, port, aktiflik durumu) değerlendirmemiz gerekiyor, ancak gelen veriler bazen hiç tanımlanmamış ( None ), bazen tanımlanmış ama boş/sıfır ( "" , 0 ) olabiliyor.
Gereksinimler:
1. yapilandirma_denetle(deger) adında bir fonksiyon oluştur.
2. Değerin hiç tanımlanıp tanımlanmadığını kontrol etmek için Python'ın önerilen yöntemi olan is None (kimlik/identity) kontrolünü kullan ve None gelirse "Ayar tanımlanmamış (None), varsayılan kullanılacak." döndür.
3. Değer tanımlanmış ( None değil) ancak Python'ın "truthiness" (doğruluk değeri) kurallarına göre "falsy" (yanlış) kabul ediliyorsa (örneğin 0 , "" , False ), if not deger: yapısını kullanarak bunu yakala ve "Değer sağlandı ancak pasif/boş." döndür.
4. Değer "truthy" (doğru) kabul ediliyorsa, "Ayar aktif: {deger}" döndür.
5. Fonksiyonu sırasıyla None , boş metin "" , 0 , False ve dolu bir metin olan "192.168.1.5" değerleriyle test et.

```python
def yapilandirme_denetle(deger):
    if deger is None:       # Bellekte tek ve eşsiz bir None nesnesi var. None olan bütün değişkenler bu aynı nesneyi referans veriyor.
        return "Ayar tanımlanmamış(None), varsayılan kullanılacak"
    elif not deger:
        return "Ayar tanımlanmış lakin falsy değere sahip."
    else:
        return f"Ayar aktif: {deger}"


deger_listesi = [None, "", 0, False, "192.168.1.5"]

for deger in deger_listesi:
    print(yapilandirme_denetle(deger))
```

### 4 - Tip Dönüşümü ve Dynamic Typing Kavramı (2/5)

Bağlam: Bir e-ticaret platformunda, dış bir API'den gelen ham ürün verilerini (her şeyin metin/string olarak geldiği bir format) işleyip envanter maliyetini hesaplayacak bir fonksiyon yazıyoruz.
Gereksinimler:
1. Parametre olarak fiyat_str ve stok_str (ikisi de string) alan bir hesaplama fonksiyonu yaz.
2. Python'da veri tiplerinin çalışma zamanında değişebilme özelliğini (dinamik tipleme) göstermek için; urun_kodu adlı bir değişkene önce tam sayı (int) olarak 1045 ata, ardından aynı değişkene "PRD-1045" string (metin) değerini atayarak tipini dinamik olarak değiştir.
3. Gelen string parametreleri aritmetik işlem yapabilmek için açıkça (explicitly) float ve int tiplerine dönüştür (casting).
4. Fiyat ve stok miktarını çarparak toplam değeri hesapla ve sonucu tekrar str() ile metne dönüştürerek ekrana yazdır. Edge-case: API'den stok bilgisi yerine yanlışlıkla sayıya dönüştürülemeyen bir metin (örneğin "stokta_yok") gelirse oluşacak
5. ValueError hatasını yakalayarak kullanıcıya anlamlı bir uyarı ver.

```python
def maliyet_hesapla(fiyat_str: str, stok_str: str) -> None:     
    try:
        fiyat_float = float(fiyat_str)
        stok_int = int(stok_str)
    except ValueError:
        print("Stokta yok!")
        return 
    toplam_deger = fiyat_float * stok_int
    print(str(toplam_deger))
    return 

urun_kodu = 1045 
urun_kodu = f"PRD-{urun_kodu}"
print("İşlenen Ürün:", urun_kodu)

ornek_veri = [("249.50", "120"),("249.50","stokta_yok")]
maliyet_hesapla(*ornek_veri[0])     # Geçerli Veri
maliyet_hesapla(*ornek_veri[1])     # Geçersiz Veri
```

### 5 - Liste Uygulamaları (2/5)

Bağlam: Bir finansal analiz aracında, bir hisse senedinin bir haftalık kapanış fiyatları üzerinde zaman serisi analizi yapabilmek için verileri anlamlı parçalara ayırıyoruz.
Gereksinimler:
1. Argüman olarak bir haftalık (7 günlük) fiyatları içeren bir liste (list) alan bir analiz fonksiyonu yaz.
2. Listenin ilk elemanını pozitif indeks, son elemanını ise negatif indeks kullanarak değişkene ata.
3. Veriyi hafta içi (ilk 5 gün) ve hafta sonu (son 2 gün) olacak şekilde iki ayrı listeye dilimle (slicing).
4. Zaman serisindeki trendi tersten okuyabilmek için dilimleme sözdizimini (slicing adımı olarak -1 ) kullanarak listenin tamamını ters çevir.
5. Edge-case: Boş bir liste gelmesi ihtimaline karşı indeks hatası (IndexError) almamak için erken çıkış (early return) yap.

```python
# Bu sorulardan sonra soru içinde deneme amaçlı veri vermesi talimatını da ekledim AI'a
haftalik_fiyatlar = [105.2, 106.8, 104.5, 108.0, 109.5, 110.1, 109.8]

def analiz_fonksiyonu(fiyatlar: list) -> None: 
    if not fiyatlar:
        print("Liste boş")
        return
    elif len(fiyatlar) != 7:
        print("Gün sayısı farklı")
        return
    else:
        veriler = {}
        veriler["ilk_eleman"] = fiyatlar[0]
        veriler["son_eleman"] = fiyatlar[-1]
        veriler["weekdays"] = fiyatlar[:5]   
        veriler["weekend"] = fiyatlar[5:]
        veriler["ters_cevrilmis"] = fiyatlar[::-1]

        for veri_adi, liste in veriler.items():
            print(veri_adi.replace("_", " ").capitalize(),":", liste) 
        return

analiz_fonksiyonu(haftalik_fiyatlar)
analiz_fonksiyonu([])
```

### 6 - Tuple Uygulamaları (2/5)

Bağlam: Bir coğrafi bilgi sistemi (CBS) uygulamasında, kritik altyapı tesislerinin konum verilerini (enlem, boylam) programın hiçbir yerinde yanlışlıkla değiştirilmeyecek şekilde güvenli bir yapıda saklamamız gerekiyor.
Gereksinimler:
1. (enlem, boylam) şeklinde iki adet float değer içeren bir demet (tuple) kabul eden bir fonksiyon yaz.
2. Edge-case: Gelen verinin gerçekten 2 boyutlu bir koordinat olup olmadığını len() ile kontrol et; eğer eksik veya fazla eleman varsa ValueError fırlat.
3. Demetin değişmezlik (immutability) özelliğini kanıtlamak için, demetin 0. indeksine yeni bir değer atamayı dene ve ortaya çıkacak TypeError istisnasını (exception) try/except bloğuyla yakalayarak ekrana koruma mesajı bas.
4. "Tuple unpacking" (paket açma) özelliğini kullanarak, demet içindeki değerleri enlem ve boylam isimli iki ayrı değişkene tek satırda ata ve formatlı bir metin olarak döndür.

```python
# Deneme Verisi
merkez_koordinat = (39.9207, 32.8541)       # Ankara, Kızılay
hatali_veri = (39.9207, 32.8541, 1000.5)    # Hatalı Format, Rakım eklenmiş

def koordinat_islem(koordinat: tuple[float, float]) -> None:        # type-hint olayı da önemli
    if not koordinat:                       # Edge-case 1
        raise ValueError("Tuple boş.")  
    if len(koordinat) != 2:                 # Edge-case 2
        raise ValueError("Tuple'da eksik veya fazla eleman var.")
     
    enlem, boylam = koordinat 
    print("Enlem:", enlem, "| Boylam:", boylam)

koordinat_islem(merkez_koordinat)

try: 
    koordinat_islem(hatali_veri)
except ValueError as hata_mesaji:
    print(f"Hata: {hata_mesaji}")

try:
    merkez_koordinat[1] = 42.8541
except TypeError as hata_mesaji:
    print(f"Hata: {hata_mesaji}")

help(koordinat_islem)       # type-hint testi
```

### 7 - Sözlük Uygulamaları (3/5)

Bağlam: Bir mikroservis uygulamasının yapılandırma (configuration) yöneticisini geliştiriyoruz. Sistem başlangıcında varsayılan ayarlar bir sözlükte tutuluyor, ancak kullanıcı tarafından verilen dış parametrelerle bu sözlüğün güvenli bir şekilde güncellenmesi ve işlenmesi gerekiyor.
Gereksinimler:
1. İki adet sözlük alan bir fonksiyon yaz: Biri varsayılan ayarlar, diğeri ise kullanıcının gönderdiği güncellemeler (override).
2. Sözlüğün .update() metodunu kullanarak, kullanıcı güncellemelerini varsayılan ayarların üzerine yaz.
3. Kritik bir değer olan "db_url" parametresini okumak için köşeli parantez [] yerine .get() metodunu kullan; böylece anahtar (key) yoksa KeyError almak yerine varsayılan olarak "sqlite://memory" atanmasını sağla.
4. Artık kullanılmayan geçici bir anahtarı ("temp_token") sistemden kaldırmak için .pop() metodunu kullan. Eğer anahtar zaten yoksa hata vermemesi için varsayılan bir dönüş değeri belirle.
5. Son olarak .items() metodunu kullanarak güncel sözlüğün üzerinde döngü kur ve tüm anahtar-değer yapılarını formatlı bir şekilde ekrana yazdır.

```python
 # Test verileri
sistem_ayari = {
        "host": "localhost",
        "port": 8080,
        "debug": False,
        "temp_token": "auth-xxx-123"
    }

yeni_ayarlar = {
        "port": 443,               # Mevcut değeri ezecek
        "debug": True,             # Mevcut değeri ezecek
        "ssl_enabled": True        # Yeni anahtar olarak eklenecek
    }

def ayarlari_guncelle(varsayilan: dict, kullanici_ayari: dict) -> dict:

    guncel_ayar = varsayilan.copy()          
    # Orijinal veriyi bozmamak için sözlük kopyası üzerinden çalışmak best-practice'miş. 
    guncel_ayar.update(kullanici_ayari)

    db_url = guncel_ayar.get("db_url", "sqlite://memory")    # Edge-case, anahtar yoksa varsayılan değer dön
    print(f"[Sistem] Veritabanı bağlantısı: '{db_url}'")

    silinen_token = guncel_ayar.pop("temp_token", None) # varsa değeri döndür ve sözlükten çıkar, yoksa ikinci argümanı döndür.
    if silinen_token:   # None ve diğer Falsy'ler kontrolü 
        print(f"[Sistem] Geçici token ('{silinen_token}') konfigürasyondan temizlendi.\n")

    return guncel_ayar
  

guncellenmis_config = ayarlari_guncelle(sistem_ayari, yeni_ayarlar)

print("--- Guncellenen Ayarlar ---")
for key, value in guncellenmis_config.items():
        print(f"{key:20s}: {value}")
```
